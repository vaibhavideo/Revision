```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load and clean
df = pd.read_csv("C:/Users/asus/Downloads/archive (2)/CardioGoodFitness.csv")
df.columns = df.columns.str.strip()  # Clean column names

# Basic info
print(df.info())
print(df.describe())

# Product distribution
sns.countplot(x="Product", data=df)
plt.title("Distribution of Products")
plt.show()

# Product vs Marital Status
sns.countplot(x="Product", hue="MaritalStatus", data=df)
plt.title("Product Preference by Marital Status")
plt.show()

# Fitness vs Usage
sns.scatterplot(x="Fitness", y="Usage", hue="Product", data=df)
plt.title("Fitness vs Usage by Product")
plt.show()

# Correlation heatmap
sns.heatmap(df.corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```
